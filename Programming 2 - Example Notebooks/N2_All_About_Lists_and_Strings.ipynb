## List Methods

We are now going to talk about the **lists**, and **list methods**.

❓Wait... "what is a *method*?" I hear you ask. Well, a method refers to a function, but more than that, **a method is a function associated with a specific data type**. For example, list methods are the functions that can be only applied to a `list` type variable, but not others.

🤓 Let me bet, you may have been famililar with some - for example, `append()` is **extremely** useful for adding a new item to the end of an existing list - you will like it!

---

For the syntax, a method comes with a dot (`.`). For example, `.append()`, `.insert()`. When calling, it is applied to `list` type variable, such as `myList.append()`, `myList.insert()`.

Some methods work with indices (positions); others work with values. **Remember: Python counts positions from 0!**

| **Method**                | **Description** |
|---------------------------|-----------------|
| `append(item)`            | Adds *item* to the end of the list. |
| `index(item)`             | Returns the index of the first element whose value is equal to *item*. A `ValueError` exception is raised if *item* is not found in the list. |
| `insert(index, item)`     | Inserts *item* into the list at the specified *index*. |
| `sort()`                  | Sorts the items in the list so they appear in ascending order (from the lowest value to the highest value). |
| `remove(item)`            | Removes the first occurrence of *item* from the list. A `ValueError` exception is raised if *item* is not found in the list. |
| `reverse()`               | Reverses the order of the items in the list. |

💡 Watch out: `index()` returns a position. The other methods above modify the original list and return `None`. Write `my_list.sort()`, not `my_list = my_list.sort()`- otherwise, your list becomes `None`. Oops.

---

#### Let's get some fruits first... 🥝

```python
# I went into a fruit shop in South Kensington...
# "Boss! What have we got today?"
fruits = ['banana', 'apple' , 'cherry']

# It's autumn - boss just gots some seasonal deli in stock!
fruits.append('pear')
print("After append:", fruits)  # [🍌, 🍎, 🍒] → [🍌, 🍎, 🍒, 🍐]

# Where are the pears...?
pear_index = fruits.index('pear')
print("Index of pear:", pear_index)

# Also some blueberries - freshly picked from Morocco at aisle 2!
fruits.insert(2, 'blueberry')
print("After insert:", fruits)  # [🍌, 🍎, 🍒, 🍐] → [🍌, 🍎, 🫐, 🍒, 🍐]

# The boss likes things alphabetical. Let's tidy up.
fruits.sort()
print("After sort:", fruits)    # [🍌, 🍎, 🫐, 🍒, 🍐] → [🍎, 🍌, 🫐, 🍒, 🍐]

# Oof... out of stock now
fruits.remove('cherry')
print("After remove:", fruits)  # [🍎, 🍌, 🫐, 🍒, 🍐] → [🍎, 🍌, 🫐, 🍐]

# New boss. Just wants everything the other way round
fruits.reverse()
print("After reverse:", fruits) # [🍎, 🍌, 🫐, 🍐] → [🍐, 🫐, 🍌, 🍎]
```

I grabbed some fruits randomly, but... have I got my five-a-day? Use `len()` to check:

```python
if len(fruits) > 5:
  print("Yum... Super healthy me! :-)")
elif len(fruits) == 5:
  print("Just right...")
else:
  print("Nooooo! :-( Must head back to the fruit shop now!")
```

To pair an element in a list with its index - there are three useful functions that massivley accelerate this process:

| **Function** | **Purpose** |
|---|---|
|`range(start, stop, step)` | Produces a sequence of integers, excluding `stop`. `start` defaults to `0` and `step` to `1`. |
| `enumerate(iterable, start=0)` | Returns each item together with a counter, starting at `0` by default. |
| `zip(iterable1, iterable2, ...)` | Groups corresponding items from multiple iterables. Stops at the shortest iterable by default. |

```python
# using range()
for i in range(len(fruits)):
    print(fruits[i],"is in aisle",i)
print()

# using enumerate()
for i, fruit in enumerate(fruits):
    print(fruits[i],"is in aisle",i)
print()

# using zip()
for i in zip(range(len(fruits)), fruits):
  print(i[1],"is in aisle",i[0])

print("Aha, my free banana is coming.")
```

## String Methods

We are now going to talk about the strings and string methods.

👍So... yes! String methods are the functions to perform common string operations such as *splitting*, *trimming*, *searching*, or *replacing*. You may find these methods particularly useful when *e.g.* clean or transform text data effectively.


| **Method** | **Description** |
|---|---|
| `startswith(substring)` | Returns `True` if the string starts with `substring`; otherwise, `False`. |
| `endswith(substring)` | Returns `True` if the string ends with `substring`; otherwise, `False`. |
| `count(substring)` | Counts non-overlapping occurrences of `substring`. |
| `find(substring)` | Returns the character index of the first occurrence of `substring`, or `-1` if not found. |
| `replace(old, new)` | Returns a new string with all occurrences of `old` replaced by `new`. |
| `lstrip(chars)` | Removes leading characters that appear in `chars`. Removes whitespace if `chars` is omitted. |
| `rstrip(chars)` | Removes trailing characters that appear in `chars`. Removes whitespace if `chars` is omitted. |
| `split(delimiter)` | Splits the string into a list using `delimiter`. If omitted, splits on runs of whitespace. |
| `separator.join(items)` | Joins strings from a list or other iterable into one string, with `separator` between them. |
| `upper()` | Returns a new string with all letters converted to uppercase. |

#### 🪟 The next day, the boss spots you walking past: "Hey, Python expert! Can you make a sign for the shop window?"

```python
# "Can you turn that list into a sign for the window?"
fruits = ['apple', 'banana', 'blueberry', 'date']

# let's try...
fruit_sign = ''.join(fruits)  # Join list items into one string
print("My first fruit sign design:", fruit_sign)

# not good... let's try with a bit trick?
fruit_sign = ', '.join(fruits)
print("My second fruit sign design:",fruit_sign)

# How many times does the letter 'a' appear?
print("Number of a's:", fruit_sign.count('a'))

# Where does 'banana' start on the sign?
print("Banana starts at:", fruit_sign.find('banana'))

# Any cherries left?
print("Cherry starts at:", fruit_sign.find('cherry'))  # -1: not found

# The boss wants the sign to be LOUDER.
print(fruit_sign.upper())

# Back to a list - because the boss has changed their mind again.
print(fruit_sign.split(', '))
```
