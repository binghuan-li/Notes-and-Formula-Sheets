## Variables and data types

Python is **dynamically typed**: an object's type comes from its value rather than a declaration. Inspect it with `type()`. Names are case-sensitive; use descriptive names that include units where helpful.

```python
sample_count = 6                 # int: a whole number
temperature_c = 36.5             # float: a decimal number
tissue_name = "cardiac muscle"   # str: text
measurements = [6.1, 6.3, 6.2]   # list: an ordered, mutable collection
is_calibrated = True             # bool: True or False
missing_value = None             # NoneType: deliberately no value

for value in [sample_count, temperature_c, tissue_name,
              measurements, is_calibrated, missing_value]:
    # print(f"{value!r:28} -> {type(value).__name__}")
    print(value, "is", type(value))
```

### Type conversion

Use `int()`, `float()`, `str()` and `bool()` to convert values. Conversion may lose information: `int(6.9)` gives `6`, whereas `round(6.9)` gives `7`.

```python
# A value read from input or a text file will often be a string.
heart_rate_text = "72"
heart_rate_bpm = int(heart_rate_text)

print(type(heart_rate_text), heart_rate_text)
print(type(heart_rate_bpm), heart_rate_bpm)
print("int(6.9) =", int(6.9))
print("round(6.9) =", round(6.9))
```

## Operators and units

`=` assigns a value; `==` compares values. Python provides `+`, `-`, `*`, `/`, `//` (floor division), `%` (remainder) and `**` (power), as well as comparisons such as `<`, `>=` and `!=`.

Python stores numbers, not their physical meaning. Include units in names and convert them explicitly. Use parentheses to make calculations clear.

```python
# Mean arterial pressure: a simplified resting approximation.
systolic_mmhg = 120
diastolic_mmhg = 75
mean_arterial_pressure_mmhg = (systolic_mmhg + 2 * diastolic_mmhg) / 3

print(f"Estimated MAP: {mean_arterial_pressure_mmhg:.1f} mmHg")
print("Is MAP at least 70 mmHg?", mean_arterial_pressure_mmhg >= 70)
```

## Strings and collections

Strings store text. An f-string allows values to be inserted directly into formatted text.

A collection stores multiple values in a single object. Common Python collections include lists, tuples, and dictionaries:
- A **list** is ordered and mutable: `[1, 2, 3]`.
- A **tuple** is ordered and immutable: `(1, 2, 3)`.
- A **dictionary** maps keys to values: `{"heart_rate": 72}`.

Here,
- Mutable means their contents can be changed after creation. 
- Immutable means their contents cannot be changed after creation.

```python
# A string stores text
patient_name = "Alex"
print(f"Patient: {patient_name}")

# A list is ordered and mutable
heart_rates = [72, 75, 74]
heart_rates.append(76)
print(f"Heart rates: {heart_rates}")

# A tuple is ordered and immutable
sample_position_mm = (12.5, 8.0, 3.5)
print(f"Sample position: {sample_position_mm}")

# A dictionary maps keys to values and is mutable
patient = {
    "id": "P014",
    "heart_rate_bpm": 72,
}

patient["heart_rate_bpm"] = 75
print(f"Patient record: {patient}")
```

### Indexing and slicing

Python uses zero-based indexing, so the first item is `[0]`. Negative indices count from the end. In `[start:stop]`, `start` is included and `stop` is excluded.

```python
glucose_mmol_per_l = [4.8, 5.1, 5.0, 5.4, 5.2]

print("First:", glucose_mmol_per_l[0])
print("Last:", glucose_mmol_per_l[-1])
print("Middle three:", glucose_mmol_per_l[1:4])
print("Number of readings:", len(glucose_mmol_per_l))
```

### Mutability

Lists and dictionaries can change. Assignment creates another name for the same object; it does not copy it. Use `.copy()` when a simple independent copy is required.

```python
original = [1.0, 2.0, 3.0]
alias = original
alias[0] = 99.0
print("Original after changing alias:", original)

independent = original.copy()
independent[0] = -1.0
print("Original after changing copy: ", original)
print("Independent copy:           ", independent)
```

## Errors are information

Read the final line of a traceback first: it provides the exception type and message. Then work upwards to find the relevant line in your code.

| Exception | Typical cause |
|---|---|
| `NameError` | using a name that has not been defined |
| `TypeError` | applying an operation to an unsuitable type |
| `ValueError` | supplying a value in an unsuitable form |
| `IndexError` | requesting a sequence position that does not exist |
| `KeyError` | requesting a dictionary key that does not exist |
| `ZeroDivisionError` | dividing by zero |

```python
measurement_text = "not recorded"

try:
    measurement = float(measurement_text)
except ValueError as error:
    print(type(error).__name__ + ":", error)
```
